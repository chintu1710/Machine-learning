```python
from matplotlib.backends.backend_pdf import PdfPages
from matplotlib import pyplot as plt
import matplotlib.patches as patches

def create_diagonal_scale_figure():
    fig, ax = plt.subplots(figsize=(11, 3))

    RF = 1 / 32
    actual_length_cm = 400  # 4 meters = 400 cm
    scale_length = RF * actual_length_cm  # = 12.5 cm on paper

    ax.plot([0, scale_length], [0, 0], color='black')

    for i in range(5):
        x = i * (scale_length / 4)
        ax.plot([x, x], [0, 1], color='black')
        ax.text(x, -0.3, f'{i}', ha='center', fontsize=9)

    sub_div_width = scale_length / 4 / 10
    for i in range(11):
        x = i * sub_div_width
        ax.plot([x, x], [0, 0.5], color='black')
        if i != 0:
            ax.text(x, 0.6, f'{i}', ha='center', fontsize=7)

    rect_height = 2.5
    ax.add_patch(patches.Rectangle((0, 0), sub_div_width, rect_height, fill=False))

    for i in range(10):
        x0 = i * sub_div_width
        x1 = (i + 1) * sub_div_width
        ax.plot([x0, x1], [0, rect_height], color='grey', linestyle='--')

    base_x = 2 * (scale_length / 4)
    plus_0_4 = 4 * sub_div_width
    x_start = base_x + plus_0_4
    diag_y = rect_height * 0.5

    ax.plot(x_start, diag_y, marker='o', color='red')
    ax.text(x_start + 0.2, diag_y + 0.2, '2.45 m', color='red', fontsize=10)

    ax.set_title('Diagonal Scale (RF = 1/32) to Measure up to 4 m – Marking 2.45 m', fontsize=11)
    ax.axis('off')
    ax.set_xlim(-1, scale_length + 1)
    ax.set_ylim(-1, rect_height + 1)

    return fig

# Save as PDF
with PdfPages("Diagonal_Scale_RF_1_32_Showing_2.45m.pdf") as pdf:
    fig = create_diagonal_scale_figure()
    pdf.savefig(fig)
    plt.close(fig)

print("PDF generated: Diagonal_Scale_RF_1_32_Showing_2.45m.pdf")
```
