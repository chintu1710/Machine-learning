What is Pandas ?
1. Pandas is a Powerful Python data analysis tolkit.
2. It is opensource
3. A fast efficient DataFrame object for data manipulation.
4. It can read and convert many file formats like csv, tsv, xml, json, excel, sql, html(webscraping) etc.
5. It is mainly used for data preprocessing.

Pandas can read in 3 diff. way
1. Series - labeled homogenous 1d
2. DataFrame - 2-d dimensional, structured, heterogeneous(diff column)
3. Panel - 3-d data

# Data wrangling, data transformation, data reduction, feature selection, and feature scaling 

# Handling of nan values

```python
# 1. Drop the nan values (Drop Nan when nan data is less than 2%)
# Try to check it or not

 # (6/12)*100
df = df.dropna()

df
```

# Method 1
# drop the rows of which are having nan values

```python
df = pd.read_csv('house.csv', na_values=['missing', 'not say', 'not found'])

df.dropna(inplace = True) # it will update the currect df
df
```

# Method 2
# Fill the nan values with same value in all columns irrespective of columns

# very less effective method of filling nan values 

```python
# 2. fill the nan values with appropriate value
df = pd.read_csv('house.csv', na_values=['missing', 'not say', 'not found'])
df = df.fillna('5555555') # it will fill all the nan value irrespective of column
# df.fillna('5555555', inplace = True)
df
```

# Method 3. 
# fill the nan value with repect to column with dictionary
# Most useful method to deal with nan values
# Calculation of values may be based on ( mean, mode, median)
# also based on some specific conditions

```python
df = pd.read_csv('house.csv', na_values=['not say', 'missing','not found'])
df
```

```python
df = df.fillna({'Build_area':700, 
                'floor':0, 
                'loan_facility':1,
               'furniture_sts': 0})
df
```
