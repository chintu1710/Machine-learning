```python
# import important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the dataset
df = pd.read_csv('diabetes.csv')
df.head()
```

Few Questions arises :

min value in some columns is zero. Is it valid?

Zero as min value found in following columns :

Glucose,Blood Prseeure,skinthickness,Insulin,BMI

```python
# We create a copy of data to deal with minimum value zero in some columns
```

```python
zero_c = ['Glucose', 'BloodPressure', 'SkinThickness','Insulin','BMI']
```

```python
# We are creating a copy to make secure the data
df_copy = df.copy()
```

```python
df_copy[zero_c] = df_copy[zero_c].replace(0, np.nan)
```

```python
df_copy.isnull().sum()
```

```python
df_copy['Glucose'].mean()
```

```python
df_copy.fillna({'Glucose':122}, inplace = True)
df_copy.isnull().sum()
```

```python
# To handle or fill the null values by mean 
# we can use SimpleImpute of Sklearn 
from sklearn.impute import SimpleImputer
im = SimpleImputer(strategy = 'mean')
```

```python
imdf = im.fit_transform(df_copy)
type(imdf)
```

```python
imdf
```

```python
imdf = pd.DataFrame(imdf, columns=df.columns)
imdf
```

```python
# Again check the nan values
imdf.isnull().sum()
```

```python
# We will check the correlation between all features and target feature
sns.heatmap(imdf.corr(), annot = True, cmap = 'Reds', fmt = '.2f')
```

```python
sns.heatmap(imdf.corr()[['Outcome']].sort_values(by = 'Outcome', ascending = False), annot = True, cmap = 'Blues') # Sort in descending order
```

```python
# Split the data into x(Independent) and y(dependent) = Outcome
x = imdf.drop(['Outcome'], axis = 1)
y = imdf['Outcome']
```

```python
x.loc[0]
```

```python
# Scale the Data
# Why we need to scale the data
# to convert the data into one scale or in a standard scale
```

```python
# So we will import standard scaler
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
x_sc = sc.fit_transform(x)
x_sc[0]
```

```python
s = sc.inverse_transform([x_sc[0]])
s
```

```python
# Now divide the data into training and testing data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_sc,y, test_size = 0.2, random_state=185,
                                                    stratify = y)
```

```python
print(x_train.shape, x_test.shape)
print(y_train.shape, y_test.shape)
```

# Import Model

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
# now train the model
knn.fit(x_train,y_train)
```

```python
# Training Score of model
knn.score(x_train, y_train) # Low 
```

```python
100-87
```

```python
y_pred = knn.predict(x_test)
```

```python
# Model Evaluation on the basis of confusion matrix, accuracy score, classification report
# import of evalation matrix
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
def model_evaluation(y_actual, y_prediction):
    print('Accuracy_Score = ',accuracy_score(y_actual, y_prediction))
    print('Confusion Matrix \n',confusion_matrix(y_actual, y_prediction))
    sns.heatmap(confusion_matrix(y_actual, y_prediction), annot = True, fmt = '.2f')
    print('Classification_Report \n',classification_report(y_actual, y_prediction))
```

```python
model_evaluation(y_test, y_pred)
```

```python
87-71
```

# Improve KNN model by changing values of K
# Find best K value

```python
er = []
test_acc = []
model_acc = []

for i in range(3,72,2):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(x_train,y_train)
    y_pred = knn.predict(x_test)
    er.append(np.mean(y_test != y_pred))
    test_acc.append(accuracy_score(y_test, y_pred))
    model_acc.append(knn.score(x_train, y_train))
```

```python
# to check the lowest error rate
plt.figure(figsize = (10,8))
plt.plot(range(3,72,2), er, marker = 'o')
plt.title('Error V/s K_value')
plt.xlabel('K Value')
plt.ylabel('Error')
plt.xticks(range(3,62,2))
plt.yticks(er)
plt.show()
```

```python
# to check the highest Accuracy
plt.figure(figsize = (10,8))
plt.plot(range(3,72,2), test_acc, marker = 'o')
plt.title('Accuracy V/s K_value')
plt.xlabel('K Value')
plt.ylabel('Accuracy')
plt.xticks(range(3,62,2))
plt.yticks(test_acc)
plt.grid()
plt.show()
```

```python
100-79.87
```

```python
# to check the highest Accuracy
plt.figure(figsize = (10,8))
plt.plot(range(3,72,2), model_acc, marker = 'o')
plt.title('Model Accuracy V/s K_value')
plt.xlabel('K Value')
plt.ylabel('Model Accuracy')
plt.xticks(range(3,62,2))
plt.yticks(model_acc)
plt.grid()
plt.show()
```
