# **Principal Component Analysis**
# Principal Component Analysis is an unsupervised learning algorithm that is used for the **dimensionality reduction** in machine learning
# It is a statistical process that converts the **observations of correlated features into a set of linearly uncorrelated** features with the help of **orthogonal transformation**.

# These new transformed features are called the **Principal Components**. It is one of the popular tools that is used for **exploratory data analysis and predictive modeling**. It is a technique to draw strong patterns from the given dataset by reducing the variances.

# **PCA generally tries to find the lower-dimensional surface to project the high-dimensional data.**

# The transformed new features or the output of PCA are the **Principal Components**. The number of these PCs are either **equal to or less than the original features** present in the dataset.

```python
ex_df = pd.read_csv('example01.csv')
print(ex_df.to_string())
print(ex_df.info())
```

```python
print(ex_df['F_mean'])
ex_df['F_mean'] = ex_df['F_mean'].astype('int64')
print(type(ex_df['F_mean']))
print(ex_df.info())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
scaler.fit(ex_df)
```

```python
print(ex_df.to_string())
```

```python
scaled_data=scaler.transform(ex_df)
```

```python
print(scaled_data)
scaled_data.shape
```

```python
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=1)
```

```python
pca.fit(scaled_data)
```

```python
x_pca=pca.transform(scaled_data)
```

```python
print(scaled_data.shape)
print(x_pca.shape)
```
